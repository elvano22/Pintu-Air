```python
import pandas as pd
import xgboost as xgb

# =========================
# LOAD DATA
# =========================
X_train = pd.read_csv("06_X_train_significant.csv")
X_test  = pd.read_csv("06_X_test_significant.csv")

y_train = pd.read_csv("04_y_train.csv")
y_test  = pd.read_csv("04_y_test.csv")

# =========================
# DROP TANGGAL DI X (KALAU ADA)
# =========================
if "Tanggal" in X_train.columns:
    X_train = X_train.drop(columns=["Tanggal"])
    X_test  = X_test.drop(columns=["Tanggal"])

# =========================
# LOAD MODEL
# =========================
model = xgb.XGBRegressor()
model.load_model("../models/13_best_model.json")

# =========================
# PREDICT
# =========================
y_train_pred = model.predict(X_train)
y_test_pred  = model.predict(X_test)

# =========================
# SAVE TRAIN PREDICT
# =========================
train_pred = pd.DataFrame({
    "Tanggal": y_train["Tanggal"],
    "y_actual": y_train["manggarai_air"],
    "y_pred": y_train_pred
})

# =========================
# SAVE TEST PREDICT
# =========================
test_pred = pd.DataFrame({
    "Tanggal": y_test["Tanggal"],
    "y_actual": y_test["manggarai_air"],
    "y_pred": y_test_pred
})

train_pred.to_csv("90_train_predict.csv", index=False)
test_pred.to_csv("90_test_predict.csv", index=False)
```
